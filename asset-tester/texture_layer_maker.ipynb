```python
import matplotlib.pyplot as plt
from PIL import Image, ImageColor
from pathlib import Path
import os
import colorsys
```

```python
def imshow(img: Image, factor: int = 20):
    from IPython.display import display
    display(img.resize((img.width * factor, img.height * factor), Image.NEAREST))
```

```python
textures_path = Path("../src/main/resources/assets/modulus/textures")
item_textures_path = textures_path/"item"
block_textures_path = textures_path/"block"

if not textures_path.exists():
    textures_path.mkdir()
if not item_textures_path.exists():
    item_textures_path.mkdir()
if not block_textures_path.exists():
    block_textures_path.mkdir()
```

```python
item_textures = list(item_textures_path.glob("*/*.png"))

item_textures
```

```python
palettes_path = Path("palettes")

if not palettes_path.exists():
    palettes_path.mkdir()
```

```python
import shutil

examples_path = Path("examples")

if examples_path.exists():
    shutil.rmtree(examples_path)

examples_path.mkdir()
```

```python
palettes = list(palettes_path.glob("*.png"))

palettes
```

```python
default_palette_path = palettes[palettes.index(Path("palettes/default.png"))]
diamond_palette_path = palettes[palettes.index(Path("palettes/diamond.png"))]

print(default_palette_path, diamond_palette_path)
```

```python
default_palette: Image = Image.open(default_palette_path)
diamond_palette: Image = Image.open(diamond_palette_path)

imshow(default_palette)
imshow(diamond_palette)
```

```python
default_palette_colors = sorted(set(c[1] for c in default_palette.getcolors()), reverse=True)

default_palette_colors
```

```python
diamond_palette_colors = sorted(set(c[1] for c in diamond_palette.getcolors()), reverse=True)

diamond_palette_colors
```

```python
tool_types = list(item_textures_path.glob("*"))

tool_types
```

```python
from copy import deepcopy

for tool_type in tool_types:
    for texture in tool_type.glob("*.png"):
        texture_dir = tool_type/texture.stem
        if not texture_dir.exists():
            texture_dir.mkdir()

        texture_img = Image.open(texture).convert("RGBA")
        layers = [Image.new("RGBA", texture_img.size, (0, 0, 0, 0)) for _ in range(len(default_palette_colors))]
        tinted_layers = deepcopy(layers)
        colored_texture_img = deepcopy(texture_img)
        
        for x in range(texture_img.width):
            for y in range(texture_img.height):
                r, g, b, a = texture_img.getpixel((x, y))    
                if a == 0:
                    continue
            
                idx = default_palette_colors.index((r, g, b))
                
                colored_texture_img.putpixel((x, y), diamond_palette_colors[idx] + (a,))
                layers[idx].putpixel((x, y), (255, 255, 255, a))
                
                rf = layers[idx].getpixel((x, y))[0] / 255
                gf = layers[idx].getpixel((x, y))[1] / 255
                bf = layers[idx].getpixel((x, y))[2] / 255
                af = layers[idx].getpixel((x, y))[3] / 255
                
                tint_r, tint_g, tint_b = diamond_palette_colors[idx]
                tint_rf = tint_r / 255
                tint_gf = tint_g / 255
                tint_bf = tint_b / 255
                tint_af = 1
                
                tinted_layers[idx].putpixel((x, y), (int(rf * tint_rf * 255), int(gf * tint_gf * 255), int(bf * tint_bf * 255), int(af * tint_af * 255)))
                
        for i, layer in enumerate(layers):
            layer.save(texture_dir/f"{i}.png")
            
        example_dir = examples_path/tool_type.stem
        if not example_dir.exists():
            example_dir.mkdir()
        
        # Composite all tinted layers together
        from functools import reduce
        tinted_img = reduce(Image.alpha_composite, tinted_layers)
        tinted_img.save(example_dir/f"{texture.stem}_tinted.png")
        colored_texture_img.save(example_dir/f"{texture.stem}_colored.png")
        
```
