```python
import matplotlib.pyplot as plt
from PIL import Image, ImageColor
from pathlib import Path
import os
import colorsys
```

```python
def imshow(img: Image, factor: int = 20):
    from IPython.display import display
    display(img.resize((img.width * factor, img.height * factor), Image.NEAREST))
```

```python
textures_path = Path("../src/main/resources/assets/modulus/textures")
item_textures_path = textures_path/"item"
block_textures_path = textures_path/"block"

if not textures_path.exists():
    textures_path.mkdir()
if not item_textures_path.exists():
    item_textures_path.mkdir()
if not block_textures_path.exists():
    block_textures_path.mkdir()
```

```python
item_textures = list(item_textures_path.glob("*/*.png"))

item_textures
```

```python
palettes_path = Path("palettes")

if not palettes_path.exists():
    palettes_path.mkdir()
```

```python
palettes = list(palettes_path.glob("*.png"))

palettes
```

```python
default_palette_path = palettes[palettes.index(Path("palettes/default.png"))]

default_palette_path
```

```python
default_palette = Image.open(default_palette_path)

imshow(default_palette)
```

```python
default_palette_colors = set(c[1] for c in default_palette.getcolors())

default_palette_colors
```

```python
for item_texture in item_textures:
    img = Image.open(item_texture).convert("RGB")
    pixels = set(img.getpixel((x, y)) for x in range(img.width) for y in range(img.height))
    
    if not pixels.issubset(default_palette_colors):
        print(f"{item_texture.parts[-2:]}: {pixels - default_palette_colors}")
```
